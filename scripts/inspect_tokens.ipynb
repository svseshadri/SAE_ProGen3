```python
import torch
from progen3.batch_preparer import ProGen3BatchPreparer

batch_preparer = ProGen3BatchPreparer()

sequence = "SESHSESH"

batch = batch_preparer.get_batch_kwargs([sequence], device="cpu")

print(batch.keys())
print("input_ids shape:", batch["input_ids"].shape)

input_ids = batch["input_ids"][0]
print("input_ids:", input_ids.tolist())

# If available
if "attention_mask" in batch:
    print("attention_mask:", batch["attention_mask"][0].tolist())
```

```python
tokenizer = batch_preparer.tokenizer  # or whatever attribute exists

for i, tok_id in enumerate(input_ids.tolist()):
    piece = tokenizer.decode([tok_id])
    print(i, tok_id, repr(piece))
```

```python
print("bos_token_id:", getattr(tokenizer, "bos_token_id", None))
print("eos_token_id:", getattr(tokenizer, "eos_token_id", None))
print("pad_token_id:", getattr(tokenizer, "pad_token_id", None))
print("cls_token_id:", getattr(tokenizer, "cls_token_id", None))
print("sep_token_id:", getattr(tokenizer, "sep_token_id", None))
```
